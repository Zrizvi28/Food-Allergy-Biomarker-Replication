```r
library(IlluminaHumanMethylationEPICanno.ilm10b4.hg19)
ann <- getAnnotation(IlluminaHumanMethylationEPICanno.ilm10b4.hg19)

isg15_probes <- rownames(ann)[ann$chr == "chr1" & ann$pos >= 948000 & ann$pos <= 950000]
rgs14_probes <- rownames(ann)[ann$chr == "chr5" & ann$pos >= 176797000 & ann$pos <= 176799000]

candidate_probes <- c(isg15_probes, rgs14_probes)
print(candidate_probes)
```

```r
data_dir <- ("/home/ethan-xiao/food-allergy-biomarkers/data")
allergy_status <- readRDS(file.path(data_dir, 'allergy_status.rds'))
batch <- readRDS(file.path(data_dir, 'batch.rds'))

# Infant cohort
mVals_114134 <- readRDS(file.path(data_dir, 'mVals_114134_shared.rds'))
infant_probes_present <- intersect(candidate_probes, rownames(mVals_114134))
infant_df <- as.data.frame(t(mVals_114134[infant_probes_present, ]))
infant_df$allergy_status <- allergy_status[batch == 'GSE114134']
write.csv(infant_df, file.path(data_dir, 'candidate_features_infant.csv'), row.names = TRUE)

# Adolescent cohort
mVals_189148 <- readRDS(file.path(data_dir, 'mVals_189148_shared.rds'))
adol_probes_present <- intersect(candidate_probes, rownames(mVals_189148))
adol_df <- as.data.frame(t(mVals_189148[adol_probes_present, ]))
adol_df$allergy_status <- allergy_status[batch == 'GSE189148']
write.csv(adol_df, file.path(data_dir, 'candidate_features_adolescent.csv'), row.names = TRUE)
```
